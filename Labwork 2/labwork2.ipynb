```python
from numba import cuda

print("=" * 60)
print("LABWORK 2 - GET TO KNOW YOUR GPU")
print("=" * 60)

# --------------------------------------------------
# 1. CUDA availability
# --------------------------------------------------
print("\n[1] CUDA availability")

if not cuda.is_available():
    print("CUDA is NOT available.")
else:
    print("CUDA is available: True")

    # --------------------------------------------------
    # 2. Detect GPU(s) - optional part of the lab
    # --------------------------------------------------
    print("\n[2] Detect GPU(s)")
    cuda.detect()

    # --------------------------------------------------
    # 3. Select GPU
    # --------------------------------------------------
    gpu_id = 0
    cuda.select_device(gpu_id)

    device = cuda.get_current_device()

    print("\n[3] Selected GPU")
    print("GPU ID:", device.id)
    print("GPU Name:", device.name)

    # --------------------------------------------------
    # 4. Core information
    # --------------------------------------------------
    print("\n[4] Core information")

    multiprocessors = device.MULTIPROCESSOR_COUNT
    print("Multiprocessor count:", multiprocessors)

    # CUDA cores per SM for common NVIDIA architectures
    # (used to calculate total CUDA cores)
    name = device.name.decode() if isinstance(device.name, bytes) else str(device.name)

    cores_per_sm = None

    if "T4" in name:
        cores_per_sm = 64
    elif "L4" in name:
        cores_per_sm = 128
    elif "A100" in name:
        cores_per_sm = 64
    elif "V100" in name:
        cores_per_sm = 64
    elif "A10" in name:
        cores_per_sm = 128
    elif "A40" in name:
        cores_per_sm = 128

    if cores_per_sm:
        total_cores = multiprocessors * cores_per_sm

        print("CUDA cores per multiprocessor:", cores_per_sm)
        print("Total CUDA core count:", total_cores)
    else:
        print("CUDA cores per multiprocessor: Check GPU architecture")
        print("Total CUDA core count: Check GPU architecture")

    # --------------------------------------------------
    # 5. Memory information
    # --------------------------------------------------
    print("\n[5] Memory information")

    free_memory, total_memory = (
        cuda.current_context().get_memory_info()
    )

    total_gb = total_memory / (1024 ** 3)
    free_gb = free_memory / (1024 ** 3)
    used_gb = total_gb - free_gb

    print(f"Total GPU memory: {total_gb:.2f} GB")
    print(f"Free GPU memory:  {free_gb:.2f} GB")
    print(f"Used GPU memory:  {used_gb:.2f} GB")

print("\n" + "=" * 60)
print("LABWORK 2 COMPLETE")
print("=" * 60)
```
